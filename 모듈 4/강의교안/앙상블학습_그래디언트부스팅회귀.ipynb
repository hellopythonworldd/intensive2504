```python
df = pd.read_csv('house_prices.csv')
df
```

```python
features = ['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 'DIS', 'RAD', 'TAX',
       'PTRATIO', 'B', 'LSTAT']

features = ['CRIM', 'ZN', 'CHAS', 'NOX', 'RM', 'AGE', 'DIS', 'RAD', 'TAX',
       'PTRATIO', 'B', 'LSTAT']

X = df[features]
y = df['MEDV']
```

```python
# 트레이닝/테스트 데이터 분할
from sklearn.model_selection import train_test_split
X_tn, X_te, y_tn, y_te = train_test_split(X, y, random_state=0)
```

```python
# 트레이닝/테스트 데이터 분할
from sklearn.model_selection import train_test_split
X_tn, X_te, y_tn, y_te = train_test_split(X, y, random_state=0)
```

```python
# 데이터 표준화
from sklearn.preprocessing import StandardScaler
std_scale = StandardScaler()
std_scale.fit(X_tn)
X_tn_std = std_scale.transform(X_tn)
X_te_std = std_scale.transform(X_te)
```

```python
# 그래디언트부스팅 학습
from sklearn.ensemble import GradientBoostingRegressor

clf_gbt = GradientBoostingRegressor(max_depth=3, learning_rate=0.1, random_state=0)
clf_gbt.fit(X_tn_std, y_tn)
```

```python
# 예측
gbt_pred = clf_gbt.predict(X_te_std)
print(gbt_pred)
```

```python
# 모형 평가-MSE
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_te, gbt_pred)
print(mse)
```
