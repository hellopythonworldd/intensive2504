## 비지도학습_DBSCAN 

```python
# 데이터 생성
from sklearn.datasets import make_moons
```

```python
X, y = make_moons(n_samples=300, noise=0.05, random_state=0)
```

```python
# 생성된 데이터 차원 확인
print(X.shape)
print(y.shape)
```

```python
# 라벨링 되지 않은 데이터 플랏
import matplotlib.pyplot as plt 

plt.scatter(X[:, 0], X[:, 1],
            c='gray', 
            edgecolor='black',
            marker='o')

plt.show()
```

```python
# DB Scan
from sklearn.cluster import DBSCAN
dbs = DBSCAN(eps=0.2)
dbs.fit(X)
label_dbs = dbs.labels_
print(label_dbs)
```

```python
# 시각화를 위한 데이터프레임 생성
import pandas as pd
dbs_columns = ['dbs_comp1', 'dbs_comp2']
X_dbs_df = pd.DataFrame(X, columns=dbs_columns)
X_dbs_df['target'] = y
X_dbs_df['label_dbs'] = label_dbs
X_dbs_df.head(5)
```

```python
# 타겟 클래스 종류 확인
print(set(X_dbs_df['target']))
print(set(X_dbs_df['label_dbs']))
```

```python
# DB scan을 이용한 데이터 플랏
df = X_dbs_df
markers=['o','x']
for i, mark in enumerate(markers):
    df_i = df[df['label_dbs']==i]
    target_i = i
    X1 = df_i['dbs_comp1']
    X2 = df_i['dbs_comp2']
    plt.scatter(X1,X2,
    marker=mark,
    label=target_i)

plt.xlabel('dbs_componentl')
plt.ylabel('dbs_component2')
plt.legend()
plt.show()
```

```python
# 실제 타겟 플랏 
df = X_dbs_df
markers=['o','x']
for i, mark in enumerate(markers):
    df_i = df[df['target']==i]
    target_i = i
    X1 = df_i['dbs_comp1']
    X2 = df_i['dbs_comp2']
    plt.scatter(X1,X2,
    marker=mark,
    label=target_i)

plt.xlabel('dbs_componentl')
plt.ylabel('dbs_component2')
plt.legend()
plt.show()
```
