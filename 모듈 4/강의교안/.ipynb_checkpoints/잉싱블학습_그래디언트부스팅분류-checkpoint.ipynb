## 앙상블학습_그래디언트부스팅분류

```python
import pandas as pd
df = pd.read_csv('wine_data.csv')
```

```python
features = ['Alcohol', 'Malic', 'Ash', 'Alcalinity', 'Magesium', 'Phenols',
       'Flavanoids', 'Nonflavanoids', 'Proanthocyanins', 'Color', 'Hue',
       'Dilution', 'Proline']

X = df[features]
y = df['class']
```

```python
# 트레이닝/테스트 데이터 분할
from sklearn.model_selection import train_test_split

X_tn, X_te, y_tn, y_te = train_test_split(X, y, random_state=0)
```

```python
# 데이터 standardization
from sklearn.preprocessing import StandardScaler
std_scale = StandardScaler()
std_scale.fit(X_tn)

X_tn_std = std_scale.transform(X_tn)
X_te_std = std_scale.transform(X_te)
```

```python
# 그래디언트부스팅 학습
from sklearn.ensemble import GradientBoostingClassifier

clf_gbt = GradientBoostingClassifier(max_depth=2, learning_rate=0.1, random_state=0)
# clf_gbt = GradientBoostingClassifier(max_depth=2, learning_rate=0.01, random_state=0)


clf_gbt.fit(X_tn_std, y_tn)
```

```python
#예측
pred_gboost = clf_gbt.predict(X_te_std)
print(pred_gboost)
```

```python
# 정밀도 (precision)
from sklearn.metrics import precision_score
precision = precision_score(y_te, pred_gboost, average='macro')
print(precision)
```

```python
# 리콜
from sklearn.metrics import recall_score
recall = recall_score(y_te, pred_gboost, average='macro')
print(recall)
```

```python
# f1 스코어
from sklearn.metrics import f1_score
f1 = f1_score(y_te, pred_gboost, average='macro')
print(f1)
```

```python
# confusion matrix 확인 
from sklearn.metrics import confusion_matrix
conf_matrix = confusion_matrix(y_te, pred_gboost)
print(conf_matrix)
```

```python
# 분류 레포트 확인
from sklearn.metrics import classification_report
class_report = classification_report(y_te, pred_gboost)
print(class_report)
```
