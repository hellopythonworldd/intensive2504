## 비지도학습_LDA

```python
import pandas as pd
df = pd.read_csv('wine_data.csv')
df
```

```python
features = ['Alcohol', 'Malic', 'Ash', 'Alcalinity', 'Magesium', 'Phenols',
       'Flavanoids', 'Nonflavanoids', 'Proanthocyanins', 'Color', 'Hue',
       'Dilution', 'Proline']

X = df[features]
y = df['class']
```

```python
# 트레이닝 /테스트 데이터 분할 
from sklearn.model_selection import train_test_split
X_tn, X_te, y_tn, y_te = train_test_split(X, y, random_state=0)
```

```python
# 데이터 스케일링
from sklearn.preprocessing import StandardScaler
std_scale = StandardScaler()
std_scale.fit(X_tn)
X_tn_std = std_scale.transform(X_tn)
X_te_std = std_scale.transform(X_te)
```

#### 지도학습으로 LDA

```python
# LDA 학습
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
lda = LinearDiscriminantAnalysis()
lda.fit(X_tn_std, y_tn)
```

```python
# 예측
pred_lda = lda.predict(X_te_std)
```

```python
# LDA 적용 후 정확도
from sklearn.metrics import accuracy_score
accuracy_lda = accuracy_score(y_te, pred_lda)
print(accuracy_lda)
```

#### LDA로 차원축소 + 지도학습

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
lda = LinearDiscriminantAnalysis()
lda.fit(X_tn_std, y_tn)
X_tn_lda = lda.transform(X_tn_std)
X_te_lda = lda.transform(X_te_std)
```

```python
# 차원축소 전후 데이터 차원 비교
print(X_tn_std.shape)
print(X_tn_lda.shape)
```
