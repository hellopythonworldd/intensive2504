## 지도학습(1)_로지스틱회귀분석 

```python
import pandas as pd 
df = pd.read_csv('wine_data.csv')
df
```

```python
df.columns
```

```python
features = ['Alcohol', 'Malic', 'Ash', 'Alcalinity', 'Magesium', 'Phenols',
       'Flavanoids', 'Nonflavanoids', 'Proanthocyanins', 'Color', 'Hue',
       'Dilution', 'Proline']

X = df[features]
y = df['class']
```

```python
# 트레이닝/테스트 데이터 분할 
from sklearn.model_selection import train_test_split
X_tn, X_te, y_tn, y_te = train_test_split(X, y, random_state=0)
```

```python
# 데이터 표준화
from sklearn.preprocessing import StandardScaler 
std_scale = StandardScaler()
std_scale.fit(X_tn)
X_tn_std = std_scale.transform(X_tn)
X_te_std = std_scale.transform(X_te)
```

```python
# 로지스틱 회귀분석 (L2 제약식 적용)
from sklearn.linear_model import LogisticRegression
clf_logi_l2 = LogisticRegression(penalty='l2')
clf_logi_l2.fit(X_tn_std, y_tn)
```

```python
# 로지스틱 회귀분석 모형 (L2 제약식 적용) 추정 계수
print(clf_logi_l2.coef_)
print(clf_logi_l2.intercept_)
```

```python
# 예측
pred_logistic = clf_logi_l2.predict(X_te_std)
print(pred_logistic)
```

```python
# 확률값으로 예측
pred_proba = clf_logi_l2.predict_proba(X_te_std)
print(pred_proba)
```

```python
# 정확도 (Accuracy)
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_te, pred_logistic)  # 실제, 예측 순서
print(accuracy)
```

```python
# 정밀도 (precision)
from sklearn.metrics import precision_score
precision = precision_score(y_te, pred_logistic, average='macro')  # 다중클래스이므로 averge 뭐로 보여줄지 결정해줘야함
print(precision)
```

```python
# 리콜 (recall)
from sklearn.metrics import recall_score
recall = recall_score(y_te, pred_logistic, average='macro')
print(recall)
```

```python
# F1 스코어
from sklearn.metrics import f1_score
f1 = f1_score(y_te, pred_logistic, average='macro')
print(f1)
```

```python
# confusion matrix 확인
from sklearn.metrics import confusion_matrix
conf_matrix = confusion_matrix(y_te, pred_logistic)
print(conf_matrix)
```

```python
# 분류 레포트 확인
from sklearn.metrics import classification_report
class_report = classification_report(y_te, pred_logistic)
print(class_report)
```
