## 지도학습(1)_KNN

```python
import pandas as pd 
```

```python
df = pd.read_csv("iris.csv")
df.head()
```

```python
df['class'].unique()
```

```python
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
X = df[features]
y = df['class']
```

```python
X
```

```python
y
```

```python
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
X_tn, X_te, y_tn, y_te = train_test_split(X, y, random_state=0, stratify=y) # todo) 숙제 stratify 적용해보기 
# X_tn, X_te, y_tn, y_te = train_test_split(X, y, random_state=0)
```

```python
y_tn.value_counts()  # stratify 옵션을 적용하면 비율이 아래와 같이 유지됨. 
```

```python
y_te.value_counts()  # stratify 옵션을 적용하면 비율이 아래와 같이 유지됨. 
```

```python
# Standardization 
from sklearn.preprocessing import StandardScaler 
std_scale = StandardScaler()  # 표준화 장치 생성 (객체)
std_scale.fit(X_tn)  # 표준화에 필요한 평균, 표준편차 계산 (X train 데이터를 기준으로)
X_tn_std = std_scale.transform(X_tn)  # 위의 평균, 표준편차로 X train 데이터 변환 
X_te_std = std_scale.transform(X_te)  # 위의 평균, 표준편차로 X test 데이터 변환 
```

```python
# 학습 
from sklearn.neighbors import KNeighborsClassifier
clf_knn = KNeighborsClassifier(n_neighbors=2)  # k=2일때가 주변 k값들보다 성능 안좋은듯?
clf_knn.fit(X_tn_std, y_tn)
```

```python
# 예측 
knn_pred = clf_knn.predict(X_te_std)
print(knn_pred)
```

```python
# 정확도 
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_te, knn_pred)
print(accuracy)
```

```python
# 정밀도 (precision)
from sklearn.metrics import precision_score
precision = precision_score(y_te, knn_pred, average='macro')
print(precision)
```

```python
# 리콜 (Recall)
from sklearn.metrics import recall_score
recall = recall_score(y_te, knn_pred, average='macro')
print(recall)
```

```python
# F1 Score
from sklearn.metrics import f1_score
f1 = f1_score(y_te, knn_pred, average='macro')
print(f1)
```

```python
# confusion matrix 확인
from sklearn.metrics import confusion_matrix
conf_matrix = confusion_matrix(y_te, knn_pred)  # 실제, 예측 순서로 데이터 들어감 
print(conf_matrix)
```

```python
# 분류 레포트 확인
from sklearn.metrics import classification_report
class_report = classification_report(y_te, knn_pred)  # 실제, 예측 순서로 데이터 들어감 
print(class_report)
```
