```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x1 = np.random.uniform(low=-10, high=10, size=100)
y = np.random.uniform(low=-10, high=10, size=100)
```

```python
plt.scatter(x1, y)
plt.show()
```

```python
x1 = np.random.uniform(low=-10, high=10, size=100)
w = 3
b = 1
y = w*x1 + b
```

```python
plt.scatter(x1, y)
plt.show()
```

```python
x1 = np.random.uniform(low=-10, high=10, size=100)
w = 3
b = 1
e = np.random.uniform(low=-10, high=10, size=100)
y = w*x1 + b + e
```

```python
plt.scatter(x1, y)
plt.show()
```
