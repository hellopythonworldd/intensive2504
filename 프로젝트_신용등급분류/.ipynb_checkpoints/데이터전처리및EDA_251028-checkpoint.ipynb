### 데이터 전처리 

#### 이상치 제거 

```python
df.select_dtypes(['float', 'int']).columns
```

```python
cols = list(df.select_dtypes(['float', 'int']).columns)
print(len(cols))

fig, axes = plt.subplots(4, 2, figsize=(12, 8))
axes = axes.flatten()

for i, col in enumerate(cols):
    sns.boxplot(data=df, y = col, ax=axes[i], color='cornflowerblue')
    axes[i].set_title(f'{col} Boxplot')

plt.tight_layout()
plt.show()
```
