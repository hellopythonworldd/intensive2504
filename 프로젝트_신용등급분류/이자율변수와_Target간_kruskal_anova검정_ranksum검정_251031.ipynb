### 이자율 변수와 loan_grade 간 N집단 평균 차이 검정 및 두 집단 간 평균 차이 검정

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns 
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("./credit_risk_dataset.csv") 
df.head() 
```

```python
df = df[['loan_int_rate', 'loan_grade']]
```

```python
df.head()
```

```python
df = df.dropna(axis=0)
```

```python
df.loan_grade.value_counts()
```

```python
df_a = df[df['loan_grade'] == 'A']['loan_int_rate']
df_b = df[df['loan_grade'] == 'B']['loan_int_rate']
df_c = df[df['loan_grade'] == 'C']['loan_int_rate']
df_d = df[df['loan_grade'] == 'D']['loan_int_rate']
df_e = df[df['loan_grade'] == 'E']['loan_int_rate']
df_f = df[df['loan_grade'] == 'F']['loan_int_rate']
df_g = df[df['loan_grade'] == 'G']['loan_int_rate']
```

```python
df_a
```

## N 집단 평균 차이 검정

#### 정규성 검정 - 샤피로 윌크 테스트 

```python
import scipy as sp
```

```python
test_stat, p = sp.stats.shapiro(df_a)  
print("검정통계량: {}, p-value: {}".format(test_stat, p)) 
print("귀무가설 기각. 정규분포를 따르지 않는다.")  
```

```python
p < 0.05
```

#### Kruskal-Wallis 테스트 

```python
test_stat, p_val = sp.stats.kruskal(df_a, df_b, df_c, df_d, df_e, df_f, df_g)  
print("검정통계량: {}, p-value: {}".format(test_stat, p_val))  
print("귀무가설 기각. 집단 간 평균에 차이가 있다.") 
```

```python
print(p_val < 0.05)
```

```python
print("집단 f 빼고 다 정규분포를 따르지 않기 때문에, rank-sum test 해야함 ")
print("7C2는 21번임.")
```

```python
df_list = [df_a, df_b, df_c, df_d, df_e, df_f, df_g]
grade_list = ['a', 'b', 'c', 'd', 'e', 'f', 'g']
```

```python
h1_counts = 0

for idx_i, grade_i in enumerate(grade_list):
    for idx_j, grade_j in enumerate(grade_list):
        if idx_i < idx_j:
            if np.mean(df_list[idx_i]) >= np.mean(df_list[idx_j]):
                test_stat, p = sp.stats.ranksums(df_list[idx_i], df_list[idx_j], alternative='greater')
                print("{},{}: 검정통계량: {}, p-value: {}".format(grade_i, grade_j, test_stat, p))
                if p < 0.05: h1_counts = h1_counts+1
            else:
                test_stat, p = sp.stats.ranksums(df_list[idx_i], df_list[idx_j], alternative='less')
                print("{},{}: 검정통계량: {}, p-value: {}".format(grade_i, grade_j, test_stat, p))
                if p < 0.05: h1_counts = h1_counts+1
```
