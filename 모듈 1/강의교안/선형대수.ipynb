#### concatenate

#### 대각화 

```python
X = df[['CRIM', 'NOX', 'RM', 'AGE', 'TAX']]
X
```

```python
X_t = X.T
```

```python
cov_x = X_t.dot(X)  
cov_x
```

```python
e, v = np.linalg.eig(cov_x)
e
```

```python
v
```

```python
np.round(v,2)
```
