#### 기초수학

```python
# min, argmin

x = [4, 2, 8, 9, 3]
min_value = min(x)
print(min_value)
```

```python
argmin_value = x.index(min_value)
print(argmin_value)
```

```python
import numpy as np
min_value = np.min(x)
print(min_value)
```

```python
argmin_value = np.argmin(x)
print(argmin_value)
```

```python
# max, argmax

x = [4, 2, 8, 9, 3]
max_value = max(x)
print(max_value)
```

```python
argmax_value = x.index(max_value)
print(argmax_value)
```

```python
max_value = np.max(x)
print(max_value)
```

```python
argmax_value = np.argmax(x)
print(argmax_value)
```

```python
import pandas as pd 
df = pd.read_csv('./data/house_prices.csv')
```

```python
df.head()
```

```python
X = df[['CRIM', 'NOX', 'RM', 'AGE', 'TAX']]
X
```

```python
X_t = np.transpose(X)
X_t
```

```python
X_t = X.T
X_t
```

```python
X_t.dot(X)
```

```python
X_t@X
```

```python
print("행렬식(determinant) 구하기")
XtX = X_t@X
np.linalg.det(XtX)
```

```python
B = df[['CRIM', 'NOX']]
B['NOX3'] = B['NOX']*3
```

```python
B
```

```python
BtB = B.T@B
np.round(np.linalg.det(BtB))
```

```python
XtX_inv = np.linalg.inv(XtX)
print(XtX_inv)
```

```python
np.round(XtX@XtX_inv)
```

```python
np.round(XtX_inv@XtX)
```

```python
crim = df['CRIM']
nox = df['NOX']
```

```python
res = np.inner(crim, nox)
print(res)
```

```python
# 대각화
cov_x = X_t.dot(X)
cov_x
```

```python
e, v = np.linalg.eig(cov_x)
```

```python
e
```

```python
np.round(v, 2)
```

```python
# A = PDP_t
print("고유값, 고유벡터로 원데이터 A 복원")
A = v@np.diag(e)@v.T
pd.DataFrame(A)
```
