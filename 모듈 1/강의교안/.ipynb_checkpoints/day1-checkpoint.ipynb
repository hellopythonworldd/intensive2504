```python
res = 0
for i in range(1,6):
    res = res + (i**2)
    print(f"{i}번째, 값 {res}")
print(res)
```

```python
res = 1
for i in range(1, 11):  # 1부터 10까지 
    res = res * i
    print(f"{i}, 값: {res}")

print(res)
```

```python
def facto(n):
    res = 1
    for i in range(1, n + 1): # 1부터 n까지 
        res *= i
    return res
```

```python
facto(5)
```

```python
def combination(n, x):
    res = facto(n)/(facto(x)*facto(n-x))
    return res
```

```python
combination(4,2)
```
