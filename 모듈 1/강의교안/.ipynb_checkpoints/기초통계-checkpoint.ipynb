```python
import numpy as np
import pandas as pd  
```

```python
df = pd.read_csv("./data/house_prices.csv")
df.head()
```

#### 방의 개수가 7개 초과일 확률 

```python
rm = df[df['RM'] > 7]
rm
```

```python
n_rm = len(rm)
print(n_rm)
```

```python
n = len(df)
print(n)
```

```python
p = n_rm/n
print(p)
```

#### 조건부 확률 

```python
ptratio = df[df['PTRATIO'] < 15]
ptratio.head(10)
```

```python
p_ptratio = len(ptratio)/len(df)
p_ptratio
```

```python
ptratio_rm = df[(df['RM'] > 7) & (df['PTRATIO'] < 15)]
ptratio_rm.head(10)
```

```python
p_ptratio_rm = len(ptratio_rm)/len(df)
p_ptratio_rm
```

```python
p = p_ptratio_rm / p_ptratio
print(p)
```

#### 평균  

```python
np.mean(df['RM'])
```

```python
df['RM'].mean()
```

```python
n = len(df)
sum_x = 0
for x in df['RM']:
    sum_x += x
mean_rm = sum_x/n
print(mean_rm)
```

#### 분산, 표준편차  

- numpy 

```python
np.var(df['RM'])
```

```python
np.std(df['RM'])
```

- pandas 

```python
df['RM'].var()
```

```python
df['RM'].std()
```

```python
n = len(df)
ss = 0
for x in df['RM']:
    ss += (x - mean_rm)**2
```

```python
var_rm = ss/n
std_rm = var_rm**0.5
print(var_rm)
print(std_rm)
```

```python
var_rm = ss/(n-1)
std_rm = var_rm**0.5
print(var_rm)
print(std_rm)
```

```python
df['RM'].var()
```

```python
df['RM'].var(ddof=1)  
```

```python
df['RM'].var(ddof=0)  
```

#### 공분산 

```python
X = df[['CRIM', 'NOX', 'RM', 'AGE', 'TAX']]
X
```

```python
cov_x = np.cov(X, rowvar=0)
cov_x
```

```python
cov_x = np.cov(X.T) 
cov_x 
```

```python
cov_x = X.cov() 
cov_x 
```

#### 피어슨 상관계수 

```python
corr_x = np.corrcoef(X.T)
print(corr_x)
```

```python
corr_x = X.corr()
corr_x
```
