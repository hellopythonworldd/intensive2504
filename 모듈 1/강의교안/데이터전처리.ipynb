####  새로운 열 추가하기 

```python
df['PRICE_N_MILEAGE'] = df['TOTAL_PRICE'] + df['MILEAGE']
df.head()
```

```python
df.drop('PRICE_N_MILEAGE', axis=1, inplace=True)
```

```python
df.head(5)
```

```python
df['GRADE'] = ''
```

```python
df.head(5)
```

```python
# loc 안쓰고 iloc쓰면 에러남
n = len(df)

for i in range(n):
    total_price = df['TOTAL_PRICE'][i]

    if total_price < 200:
        df.loc[i, 'GRADE'] = 'NORMAL'
    elif 200 <= total_price < 500:
        df.loc[i, 'GRADE'] = 'VIP'
    else:
        df.loc[i, 'GRADE'] = 'VVIP'
```
