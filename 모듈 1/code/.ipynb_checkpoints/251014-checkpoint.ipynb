```python
import numpy as np
```

#### argmin

```python
x = [4,2,8,9,3] 
min_value = min(x)
print(min_value)
```

```python
argmin_value = x.index(min_value)
print(argmin_value)
```

```python
min_value = np.min(x)
print(min_value)
```

```python
argmin_value = np.argmin(x)
print(argmin_value)
```

#### argmax 

```python
x = [4,2,8,9,3] 
max_value = max(x)
print(max_value)
```

```python
argmax_value = x.index(max_value)
print(argmax_value)
```

```python
max_value = np.max(x)
print(max_value)
```

```python
argmax_value = np.argmax(x)
print(argmax_value)
```
