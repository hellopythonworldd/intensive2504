#### 상속 

```python
class Country01:
    name = '국가명'
    capital = '수도'
    population = 100

    def introduce(self):
        sentence = '국가 소개 문장입니다. '
        return sentence
```

```python
class Korea01(Country01):
    def __init__(self, name):
        self.name = name

    def explanation(self):
        sentence = '국가 이름은 {} 입니다.'.format(self.name)
        return sentence
```

```python
ex3 = Korea01(name='한국')
ex3.name
```

```python
ex3.capital
```

```python
ex3.population
```

```python
ex3.introduce()
```

```python
ex3.explanation()
```

#### 메소드 오버라이딩 

```python
class Country01:
    name = '국가명'
    capital = '수도'
    population = 100

    def introduce(self):
        sentence = '국가 소개 문장입니다. '
        return sentence
```

```python
class Korea02(Country01):
    def __init__(self, name):
        self.name = name

    def introduce(self):
        sentence = '국가 이름은 {} 입니다.'.format(self.name)
        return sentence 
```

```python
ex4 = Korea02(name='한국')
ex4.name
```

```python
ex4.capital
```

```python
ex4.population
```

```python
ex4.introduce()
```

```python
class Korea03(Country01):
    def __init__(self, name):
        self.name = name

    def introduce(self):
        res1 = super().introduce()
        res2 = '국가 이름은 {} 입니다.'.format(self.name)
        return res1, res2
```

```python
ex5 = Korea03(name='한국')
```

```python
ex5.name
```

```python
ex5.capital
```

```python
ex5.population
```

```python
a1, a2 = ex5.introduce()
print(a1)
print(a2)
```
