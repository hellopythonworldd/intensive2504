## 딥러닝기초회귀_복습

```python
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, random_split, DataLoader
```

```python
device = 'cpu'
if torch.cuda.is_available():
    device = 'cuda'

torch.manual_seed(777)
if device=='cuda':
    torch.cuda.manual_seed_all(777)
```

```python
df = pd.read_csv('house_prices.csv')
```

```python
df.head(3)
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
X = torch.tensor(X, dtype=torch.float32).to(device)
y = torch.tensor(y, dtype=torch.float32).to(device)
```

```python
dataset = TensorDataset(X, y)
```

```python
n = len(dataset)
ratio_tn = 0.8
n_tn = int(n*ratio_tn)
n_te = n - n_tn
```

```python
print("전체 데이터 개수: ", n)
print("트레이닝 데이터 비율: ", ratio_tn)
print("트레이닝 데이터 개수: ", n_tn)
print("테스트 데이터 개수: ", n_te)
```

```python
train_data, test_data = random_split(dataset, [n_tn, n_te])
# for i in range(0, 3):
#     feature, target = train_data[i]
#     print("Input: ", feature)
#     print("Label: ", target)
#     print()
```

```python
batch_size = 16
train_loader = DataLoader(train_data, batch_size=batch_size, shuffle=True)
```

```python
# hyperparameters
learning_rate = 0.01
epochs = 1000
n_feature = len(train_data[0][0])
print(n_feature)
```

```python
class Regressor(nn.Module):
    def __init__(self, n_feature):
        super().__init__()
        self.fc1 = nn.Linear(n_feature, 30)
        self.bn1 = torch.nn.BatchNorm1d(30)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(p=0.3)
        self.fc2 = nn.Linear(30, 1)

    def forward(self, x):
        x = self.fc1(x)
        x = self.bn1(x)
        x = self.relu(x)
        x = self.dropout(x)
        x = self.fc2(x)

        return x
```

```python
model = Regressor(n_feature).to(device)
```

```python
criterion = nn.MSELoss().to(device)
optimizer = optim.SGD(model.parameters(), lr=learning_rate)
```

```python
total_batch = len(train_loader)
model.train()
for epoch in range(epochs):
    avg_cost = 0
    for X_tn, y_tn in train_loader:

        X_tn = X_tn.to(device)
        y_tn = y_tn.to(device)

        optimizer.zero_grad()
        y_hat = model.forward(X_tn)
        y_hat = y_hat.squeeze()
        cost = criterion(y_hat, y_tn)
        cost.backward()
        optimizer.step()

        avg_cost += cost/total_batch

    if (epoch+1) % 100 == 0:
        print("Epoch [%d/%d], Cost: %.4f " % (epoch+1, epochs, avg_cost))
```

```python
test_data[0]
```

```python
X_te = []
y_te = []

for row in test_data:
    feature, target =  row
    X_te.append(feature)
    y_te.append(target)
```

```python
X_te = torch.stack(X_te)
y_te = torch.stack(y_te)
```

```python
model.eval()
with torch.no_grad():
    prediction = model.forward(X_te)
    prediction = prediction.squeeze()
    mse = criterion(prediction, y_te)
    print("MSE: ", mse.item())
```
