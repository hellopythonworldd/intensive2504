## 파이토치를활용한분류_구조심화 

```python
import pandas as pd

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, random_split, DataLoader
```

```python
device = 'cpu'
if torch.cuda.is_available():
    device = 'cuda'

torch.manual_seed(777)
if device == 'cuda':
    torch.cuda.manual_seed_all(777)
```

```python
print(device)
```

```python
df = pd.read_csv('iris.csv')
df.head()
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
X[0:10]
```

```python
y
```

```python
X = torch.tensor(X, dtype=torch.float32).to(device)
y = torch.tensor(y, dtype=torch.long).to(device)
```

```python
y
```

```python
dataset = TensorDataset(X, y)
```

```python
for i in range(0, 10):
    print(dataset[i])
```

```python
n = len(dataset)
ratio_tn = 0.8
n_tn = int(n*ratio_tn)
n_te = n - n_tn
```

```python
print("전체 데이터 개수: ", n)
print("트레이닝 데이터 비율: ", ratio_tn)
print("트레이닝 데이터 개수: ", n_tn)
print("테스트 데이터 개수: ", n_te)
```

```python
train_data, test_data = random_split(dataset, [n_tn, n_te])
```

```python
train_data[0]
```

```python
train_data[1]
```

```python
n_batch = 16
train_loader = DataLoader(train_data, batch_size=n_batch, shuffle=True)
```

```python
batch_count=0
for minibatch in train_loader:
    features, targets = minibatch
    print("Feature: ", features)
    print("Target: ", targets)
    print()

    batch_count += 1
    if batch_count > 2:
        break
```

```python
# hyperparameters
learning_rate = 0.01
epochs = 1000
n_feature = len(train_data[0][0])
print(n_feature)
```

```python
class Classifier(nn.Module):
    def __init__(self, n_feature):
        super().__init__()
        self.layer1 = nn.Sequential(
                        nn.Linear(n_feature, 20),
                        nn.BatchNorm1d(num_features=20),
                        nn.LeakyReLU(),
                        nn.Dropout(p=0.2)
                        )

        self.layer2 = nn.Sequential(
                        nn.Linear(20, 20),
                        nn.BatchNorm1d(num_features=20),
                        nn.LeakyReLU(),
                        nn.Dropout(p=0.2)
                        )
        self.layer3 = nn.Sequential(
                        nn.Linear(20, 3),
                        nn.BatchNorm1d(num_features=3)
                        )


    def forward(self, x):
        x = self.layer1(x)
        x = self.layer2(x)
        x = self.layer3(x)
        return x
```

```python
model = Classifier(n_feature).to(device)
```

```python
criterion = nn.CrossEntropyLoss().to(device)
optimizer = optim.SGD(model.parameters(), lr=learning_rate)
```

```python
total_batch = len(train_loader)
model.train() 

for epoch in range(epochs):
    avg_cost = 0
    for X_tn, y_tn in train_loader:

        X_tn = X_tn.to(device)
        y_tn = y_tn.to(device)

        optimizer.zero_grad()
        y_hat = model.forward(X_tn)
        cost = criterion(y_hat, y_tn)
        cost.backward()
        optimizer.step()

        avg_cost += cost/total_batch

    if (epoch+1)%100 == 0:
        print("Epoch [%d/%d], Cost: %.4f" % (epoch+1, epochs, avg_cost))
```

```python
y_tn
```

```python
X_te = []
y_te = []

for row in test_data:
    feature, target = row
    X_te.append(feature)
    y_te.append(target)
```

```python
X_te = torch.stack(X_te)
y_te = torch.stack(y_te)
```

```python
model.eval()
with torch.no_grad():
    prediction = model.forward(X_te)
    correct_prediction = torch.argmax(prediction, 1) == y_te
    accuracy = correct_prediction.float().mean()
    print('Accuracy: ', accuracy.item())
```
