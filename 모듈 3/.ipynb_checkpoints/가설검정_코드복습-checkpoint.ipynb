#### 가설검정_코드복습 

#### 두 집단 평균 차이 검정 

```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv("./performance.csv")
df.head()
```

```python
df['TEAM'].unique()
```

```python
df_a = df[df['TEAM'] == 'A']
df_b = df[df['TEAM'] == 'B']
df_c = df[df['TEAM'] == 'C']
df_d = df[df['TEAM'] == 'D']
df_e = df[df['TEAM'] == 'E']
```

```python
df_a_ach = df_a['ACHIEVEMENT']
df_b_ach = df_b['ACHIEVEMENT']
df_c_ach = df_c['ACHIEVEMENT']
df_d_ach = df_d['ACHIEVEMENT']
df_e_ach = df_e['ACHIEVEMENT']
```

```python
# 가설설정 
print("H0: 팀A와 팀B의 평균 성과 점수는 차이가 없다.")
print("H1: 팀B의 성과점수가 팀A의 성과점수보다 크다. ")
```

```python
# 정규성검정 (샤피로 윌크 검정)
import scipy as sp
test_stat, p = sp.stats.shapiro(df_a_ach)
print("H0: 정규 분포를 따른다. ")
print("H1: 정규 분포를 따르지 않는다.")
print("검정통계량: {}, p-value: {}".format(test_stat, p))
print("귀무가설 기각할 수 없음, 정규분포를 따른다. ")
```

```python
test_stat, p = sp.stats.shapiro(df_b_ach)
print("H0: 정규 분포를 따른다. ")
print("H1: 정규 분포를 따르지 않는다.")
print("검정통계량: {}, p-value: {}".format(test_stat, p))
print("귀무가설 기각, 정규분포를 따르지 않는다. ")
```

```python
# z-test 
df_a_mean = np.mean(df_a_ach)
df_b_mean = np.mean(df_b_ach)
print(df_a_mean)
print(df_b_mean)
```

```python
df_a_var = df_a_ach.var(ddof=0)
df_b_var = df_b_ach.var(ddof=0)
print(df_a_var)
print(df_b_var)
```

```python
len_a = len(df_a_ach)
len_b = len(df_b_ach)
print(len_a)
print(len_b)
```

```python
z_value = (df_b_mean - df_a_mean)/((df_a_var/len_a) + (df_b_var/len_b))**0.5
print(z_value)
```

```python
z_dist = sp.stats.norm(0, 1)
p_value = 1-z_dist.cdf(z_value)
print(p_value)
```

```python
print("H0: 두 평균의 차이가 없다.")
print("H1: 두 평균의 차이가 있다. ")
print("H0 기각할 수 없다. 두 평균의 차이가 없다. ")
```

```python
# t-test
test_stats, p = sp.stats.ttest_ind(df_b_ach, df_a_ach, alternative='greater')
print("검정통계량: {}, p-value: {}".format(test_stat, p))
```

```python
# ranksum tset 
test_stats, p = sp.stats.ranksums(df_b_ach, df_a_ach, alternative='greater')
print("검정통계량: {}, p-value: {}".format(test_stat, p))
```

#### 세 집단의 평균 비교 

```python
print("H0: 팀A, 팀B, 팀C, 팀D, 팀E의 평균 성과 점수는 차이가 없다.")
print("H1: 팀A, 팀B, 팀C, 팀D, 팀E의 평균 성과 점수는 차이가 있다.")
```

```python
# 정규성 검정 (shapiro)
```

```python
# 등분산성 검정 
print("H0: 팀 A,B,C,D,E의 분산이 동일하다. ")
print("H1: 팀 A,B,C,D,E의 분산이 동일하지 않다. ")
test_stat, p = sp.stats.bartlett(df_a_ach, df_b_ach, df_c_ach, df_d_ach, df_e_ach)
print("검정통계량: {}, p-value: {}".format(test_stat, p))
print("귀무가설 기각할 수 없음, 분산이 동일하다. ")
```

```python
# ANOVA 
test_stat, p = sp.stats.f_oneway(df_a_ach, df_b_ach, df_c_ach, df_d_ach, df_e_ach)
print("검정통계량: {}, p-value: {}".format(test_stat, p))
print("귀무가설 기각, 평균 성과 점수에 차이가 있다.")
```

```python
# kruskal
test_stat, p = sp.stats.kruskal(df_a_ach, df_b_ach, df_c_ach, df_d_ach, df_e_ach)
print("검정통계량: {}, p-value: {}".format(test_stat, p))
print("귀무가설 기각, 성과 점수 중앙값에 차이가 있다.")
```

#### 연관성 검정 - 카이제곱 검정 

```python
print("H0: 팀C여부와 사내 수강 여부에는 연관성이 없다. ")
print("H1: 팀C여부와 사내수강 여부에는 연관성이 있다.")
```

```python
df['C_OR_NOT'] = np.where(df['TEAM']=='C', 'C', 'not_C')
df.head()
```

```python
df['TRAINING'].unique()
```

```python
df['C_OR_NOT'].unique()
```

```python
obs = pd.crosstab(df['C_OR_NOT'], df['TRAINING'])
obs
```

```python
test_stat, p, d, expected = sp.stats.chi2_contingency(obs)
print("검정통계량: {}, p-value: {}".format(test_stat, p)) 
print("H0 기각할 수 없음, 연관성이 없다.")
```
