### 확률분포 

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

## 이산형

#### 균일분포 

```python
np.random.seed(42)
x = np.random.randint(low=1, high=6, size=100)
x
```

```python
plt.hist(x, bins=10)
plt.show()
```

```python
x = np.random.randint(low=1, high=6, size=100000)
x
```

```python
plt.hist(x, bins=10)  # size 값이 커지면 평평해짐 
plt.show()
```

### 베르누이 분포  

```python
x = np.random.binomial(n=1, p=1/3, size=100)
x
```

```python
plt.hist(x, bins=3)
plt.show()
```

```python
x = np.random.binomial(n=1, p=1/3, size=10000)
x
```

```python
pd.DataFrame(x.T).value_counts()
```

- 1이 나올 확률이 1/3이기 때문에 아래와 같은 모양이 나옴 

```python
plt.hist(x, bins=3)
plt.show()
```

### 이항분포  

```python
x = np.random.binomial(n=10, p=1/3, size=100)
x
```

```python
plt.hist(x, bins=30)
plt.show()
```

```python
x = np.random.binomial(n=10, p=1/3, size=1000000)
```

```python
plt.hist(x, bins=30)
plt.show() # 약간 왼쪽으로 치우쳐있음. 성공확률이 1/3이니까 적은편이니까 왼쪽으로 치우쳐있음 
```

```python
x.max() 
```

```python
x = np.random.binomial(n=10, p=1/2, size=100000)  
plt.hist(x, bins=30)  
plt.show()  # p = 1/2인 경우에는 거의 대칭  
```

```python
x = np.random.binomial(n=10, p=2/3, size=100000)  
plt.hist(x, bins=30)  
plt.show()  # 성공확률 2/3이니까 약간 오른쪽으로 치우쳐있음  
```

#### 포아송분포  

```python
x = np.random.poisson(lam=2, size=100)
plt.hist(x, bins=30)
plt.show()
```

```python
x = np.random.poisson(lam=2, size=1000000)
plt.hist(x, bins=30)
plt.show()
```

#### 기하분포  

```python
x = np.random.geometric(p=1/3, size=100)
plt.hist(x, bins=30)
plt.show()
```

```python
x = np.random.geometric(p=1/3, size=100000)
plt.hist(x, bins=30)
plt.show()
```

## 연속형

```python
x = np.random.uniform(low=1, high=6, size=100)
plt.hist(x, bins=30)
plt.show()
```

```python
x = np.random.uniform(low=1, high=6, size=1000000)
plt.hist(x, bins=30)
plt.show()
```

#### 정규분포 

```python
x = np.random.normal(loc=60, scale=6, size=100)
plt.hist(x, bins=30)
plt.show()
```

```python
x = np.random.normal(loc=60, scale=6, size=1000)
plt.hist(x, bins=30)
plt.show()
```

#### 감마분포  

```python
x = np.random.gamma(shape=2, scale=3, size=30)
plt.hist(x, bins=30)
plt.show()
```

```python
x = np.random.gamma(shape=2, scale=3, size=1000)
plt.hist(x, bins=30)
plt.show()
```

```python
x = np.random.gamma(shape=1, scale=10, size=1000)
plt.hist(x, bins=30)
plt.show()
```

```python
x = np.random.gamma(shape=2, scale=2, size=1000) 
plt.hist(x, bins=30) 
plt.show() 
```

```python
x = np.random.gamma(shape=10000, scale=2, size=1000)  # 궁금하면 감마분포의 정규근사 찾아보기 
plt.hist(x, bins=30) 
plt.show() 
```

#### 지수분포  

```python
x = np.random.exponential(scale=2, size=1000) 
plt.hist(x, bins=30)
plt.show() 
```

```python
x = np.random.exponential(scale=2, size=10000) 
plt.hist(x, bins=30)
plt.show() 
```

```python
x = np.random.exponential(scale=100, size=10000) 
plt.hist(x, bins=30)
plt.show() 
```

#### 카이제곱분포 

```python
x = np.random.chisquare(df=3, size=1000)
plt.hist(x, bins=30)
plt.show()
```

```python
x = np.random.chisquare(df=1000, size=1000)
plt.hist(x, bins=30)
plt.show()
```
